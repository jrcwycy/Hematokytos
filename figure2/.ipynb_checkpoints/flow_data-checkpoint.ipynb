```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import mannwhitneyu, permutation_test, shapiro, ttest_ind
```

# Load flow cytometry data

```python
path = "../results/Flow_Data.csv"

df = pd.read_csv(path)
df['CD34+ %'] = df['CD34+ %'].str.rstrip('%').astype(float)

print(df.shape)
df
```

```python
plt.rcParams['figure.dpi'] = 200
plt.rcParams['figure.figsize'] = 4, 4

fig, ax = plt.subplots()

order = df['Condition'].unique()
line_width = 1

sns.barplot(
    data=df,
    x='Condition',
    y='CD34+ %',
    hue='Condition',
    errorbar=None,
    capsize=0.2,
    width=0.4,
    ec='k',
    linewidth=1,
    ax=ax,
)

# add SEM errorbars
summary = df.groupby('Condition')['CD34+ %'].agg(['mean', 'sem']).reindex(order)
ax.errorbar(
    x=np.arange(len(order)),
    y=summary['mean'],
    yerr=[np.zeros(len(order)), summary['sem']],
    fmt='none',
    color='k',
    elinewidth=1,
    capsize=3, 
    zorder=10,
)

# overlay individual replicates
sns.stripplot(
    data=df,
    x='Condition',
    y='CD34+ %',
    marker='o',
    jitter=0.1,
    size=3,
    edgecolor='k',
    linewidth=0.5,
    facecolors='none',
    ax=ax,
)

ax.set_ylabel("% CD34+", fontsize=12) 
ax.set_xlabel("")


plt.title('CD34+ percentage across conditions')

ax.tick_params(axis='x', rotation=90, labelsize=12)
ax.tick_params(axis='y', labelsize=10)
ax.tick_params(axis='both', width=line_width)
          
for spine in ax.spines.values():
    spine.set_linewidth(line_width)
    
sns.despine()
plt.tight_layout() 
plt.show()     
```

# Compare control vs 5TF

```python
conditions = ['Control (Dox-)', 'FH4a', 'FH6']

tmp = df[df['Condition'].isin(conditions)].copy()

# rename control for plotting
tmp['name'] = np.where(
    tmp['Condition'] == 'Control (Dox-)',
    'control',
    tmp['N TF']
)

# exclude 3TF control sample
tmp = tmp[tmp['N TF'] != '3TF']

tmp
```

```python
plt.rcParams['figure.dpi'] = 200
plt.rcParams['figure.figsize'] = 1.75, 2

fig, ax = plt.subplots()

order = ['control', '5TF']

sns.barplot(
    data=tmp,
    x='name',
    y='CD34+ %',
    errorbar=None,
    capsize=0.2,
    width=0.4,
    ec='k',
    linewidth=1,
    color='white',
    ax=ax,
)

# add errorbars
summary = tmp.groupby('name')['CD34+ %'].agg(['mean', 'sem']).reindex(order)

ax.errorbar(
    x=np.arange(len(order)),
    y=summary['mean'],
    yerr=summary['sem'],
    fmt='none',
    color='k',
    elinewidth=1,
    capsize=3, 
    zorder=1,
)

# overlay replicates
sns.stripplot(
    data=tmp,
    x='name',
    y='CD34+ %',
    hue='name',
    marker='o',
    jitter=True,
    size=3,
    edgecolor='k',
    linewidth=0,
    palette=['Grey', 'deepskyblue'],
    ax=ax,
)

ax.set_ylim([0, 4.5])
ax.set_ylabel("% CD34$^+$", fontsize=12) 
ax.set_xlabel("")

ax.tick_params(axis='x', labelsize=12)
ax.tick_params(axis='y', labelsize=10)
ax.tick_params(axis='both', width=1)

for spine in ax.spines.values():
    spine.set_linewidth(1)
    
sns.despine()
plt.tight_layout() 
plt.show()     
```

# Statistical testing

```python
control = tmp.loc[tmp['name'] == 'control', 'CD34+ %']
tf5 = tmp.loc[tmp['name'] == '5TF', 'CD34+ %']

# Mann-Whitney U test
stat, p = mannwhitneyu(tf5, control, alternative='greater')
print("U =", stat, "p-value =", p)

stat, p = mannwhitneyu(tf5, control, alternative='two-sided')
print("U =", stat, "p-value =", p)
```

```python
### Assess normality
# control has only n=2, so Shapiro-Wilk is not applicable
stat1, p1 = shapiro(tf5)
print("Shapiro Wilks: stat =", stat1, "p-value =", p1)


fig, ax = plt.subplots(figsize=(4, 4))

stats.probplot(tf5, dist="norm", plot=ax)

ax.set_title("Q-Q Plot: 5TF %CD34+")
ax.set_xlabel("Theoretical Quantiles")
ax.set_ylabel("Observed Quantiles")

plt.tight_layout()
plt.show()
```

```python
### Welch's t-test
stat, p = ttest_ind(
    tf5,
    control,
    equal_var=False,
    alternative="greater"
)

print("Welch's t-test (greater): stat =", stat, "p-value =", p)

stat, p = ttest_ind(
    tf5,
    control,
    equal_var=False,
    alternative="two-sided"
)

print("Welch's t-test (two-sided): stat =", stat, "p-value =", p)
```

```python
def mean_difference(x, y):
    """Difference in group means."""
    return np.mean(x) - np.mean(y)

result = permutation_test(
    (tf5, control),
    mean_difference,
    permutation_type="independent",
    alternative="greater",
)

print(f"Permutation test p-value = {result.pvalue:.4g}")
```
